# Explore datasets
All datasets share the `Stream` format. Run `python scripts/download_data.py <name>` first.

```python
from collections import Counter
from IPython.display import display
from streamrag.data import get_dataset
from streamrag.data.view import show_stream
```

```python
streams = {name: get_dataset(name).to_streams() for name in ['disastervqa', 'videomme', 'm3bench', 'mindcube']}
for name, ss in streams.items():
    print(f"{name:12s} streams={len(ss):4d}  obs={sum(len(s.observations) for s in ss):6d}  qa={sum(len(s.qa) for s in ss):5d}")
```

```python
# question-type distribution per dataset
for name, ss in streams.items():
    c = Counter(str(q.qtype) for s in ss for q in s.qa)
    print(name, c.most_common(8))
```

```python
# look at one stream (prints timeline + QA, returns the contact sheet path)
from PIL import Image
path = show_stream(streams['m3bench'][0], max_obs=10, max_qa=3)
display(Image.open(path)) if path else None
```
