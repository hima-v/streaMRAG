# Model smoke test
Loads one model at a time (24 GB GPU). `unload()` frees VRAM between models.

```python
import torch
from streamrag.models import load_model, unload
from streamrag.data import get_dataset
s = get_dataset('disastervqa').to_streams(limit=1)[0]
img = s.observations[0].load_image(); img
```

```python
vlm = load_model('qwen2-vl-2b')
print(vlm.generate('Describe this scene for an incident log.', images=[img], max_new_tokens=80))
print(f'{torch.cuda.max_memory_allocated()/2**30:.1f} GB'); unload()
```

```python
v = get_dataset('m3bench').to_streams(limit=1)[0]
frames = [o.load_image() for o in v.observations if o.modality == 'frame'][:8]
vlm = load_model('nemotron-12b-vl')
print(vlm.generate('What happens across these frames?', frames=frames, max_new_tokens=96)); unload()
```

```python
nli = load_model('nli-deberta-large')
print(nli.predict([('The hallway is empty.', 'A person walks through the hallway.')]))
```
